```python
import glob, subprocess, sys
run = glob.glob("/kaggle/input/**/kaggle_run.py", recursive=True)[0]
print(run, flush=True)
print(sorted(set(p.split("/")[3] for p in glob.glob("/kaggle/input/*/*"))), flush=True)
common = ["--dtype", "float16", "--conditions", "closed", "--questions", "mkqa_questions.jsonl", "--kaggle-model", "/kaggle/input/**/"]
jobs = [
    ["--model", "CohereLabs/tiny-aya-global", "--tag", "t4-mkqa", "--hours", "1.5"] + common,
    ["--model", "CohereLabs/aya-expanse-8b", "--tag", "mkqa", "--hours", "2"] + common,
]
codes = [subprocess.run([sys.executable, run] + job).returncode for job in jobs]   # the second still runs if the first fails
print("exit codes", codes, flush=True)
assert not any(codes)
```
